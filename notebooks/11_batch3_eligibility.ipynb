```python
# ============================================================
# NOTEBOOK 11 — BATCH 3 ELIGIBILITY
# CELL 1 — IMPORTS AND PATHS
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# Project directories
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    r"D:\Pancreatic_Cancer_Thesis"
)

DATA_DIR = PROJECT_ROOT / "data"

RAW_CT_DIR = DATA_DIR / "raw_ct"

AUTO_LABEL_DIR = (
    DATA_DIR
    / "labels"
    / "Automatic_Labels"
)

MANUAL_LABEL_DIR = (
    DATA_DIR
    / "labels"
    / "Manual_Labels"
)

PROCESSED_DIR = (
    DATA_DIR
    / "processed"
)

BATCH3_INVENTORY_FILE = (
    PROCESSED_DIR
    / "batch3_inventory.csv"
)

BATCH3_ELIGIBILITY_FILE = (
    PROCESSED_DIR
    / "batch3_eligibility.csv"
)


print("=" * 70)
print("NOTEBOOK 11 — BATCH 3 ELIGIBILITY")
print("=" * 70)

print("\nProject root:")
print(PROJECT_ROOT)

print("\nBatch 3 inventory:")
print(BATCH3_INVENTORY_FILE)

print("\nEligibility output:")
print(BATCH3_ELIGIBILITY_FILE)
```

```python
# ============================================================
# CELL 2 — LOAD BATCH 3 INVENTORY
# ============================================================

if not BATCH3_INVENTORY_FILE.exists():
    raise FileNotFoundError(
        f"Batch 3 inventory not found:\n"
        f"{BATCH3_INVENTORY_FILE}"
    )

batch3_inventory = pd.read_csv(
    BATCH3_INVENTORY_FILE
)

print("=" * 70)
print("BATCH 3 INVENTORY LOADED")
print("=" * 70)

print(
    "Rows    :",
    len(batch3_inventory)
)

print(
    "Columns :",
    len(batch3_inventory.columns)
)

print("\nColumns:")
for column in batch3_inventory.columns:
    print(" ", column)

print("\nFirst 5 cases:")
display(
    batch3_inventory.head()
)
```

```python
# ============================================================
# CELL 3 — INVENTORY SANITY CHECK
# ============================================================

print("=" * 70)
print("BATCH 3 INVENTORY SANITY CHECK")
print("=" * 70)

expected_cases = 580

print(
    "Expected cases :",
    expected_cases
)

print(
    "Actual cases   :",
    len(batch3_inventory)
)

assert len(batch3_inventory) == expected_cases, (
    f"Expected {expected_cases} cases, "
    f"found {len(batch3_inventory)}."
)

# ------------------------------------------------------------
# Required columns
# ------------------------------------------------------------

required_columns = [
    "study_id",
    "ct_readable",
    "label_readable",
    "geometry_match",
    "selected_label_type",
]

missing_columns = [
    column
    for column in required_columns
    if column not in batch3_inventory.columns
]

if missing_columns:
    raise ValueError(
        "Missing required inventory columns:\n"
        + "\n".join(missing_columns)
    )

print("\nRequired columns present ✓")

# ------------------------------------------------------------
# Duplicate study IDs
# ------------------------------------------------------------

duplicate_ids = (
    batch3_inventory[
        batch3_inventory["study_id"].duplicated(
            keep=False
        )
    ]
)

print(
    "\nDuplicate study IDs:",
    len(duplicate_ids)
)

if len(duplicate_ids) > 0:
    display(duplicate_ids)
    raise ValueError(
        "Duplicate study IDs detected."
    )

# ------------------------------------------------------------
# Basic inspection
# ------------------------------------------------------------

print("\nCT readable:")
print(
    batch3_inventory["ct_readable"]
    .value_counts(dropna=False)
)

print("\nLabel readable:")
print(
    batch3_inventory["label_readable"]
    .value_counts(dropna=False)
)

print("\nGeometry match:")
print(
    batch3_inventory["geometry_match"]
    .value_counts(dropna=False)
)

print("\nSelected label type:")
print(
    batch3_inventory["selected_label_type"]
    .value_counts(dropna=False)
)
```

```python
# ============================================================
# CELL 4 — BUILD BATCH 3 ELIGIBILITY
# ============================================================

HELD_OUT_CASE = "100936_00001"

eligibility = batch3_inventory.copy()

# ------------------------------------------------------------
# Individual eligibility conditions
# ------------------------------------------------------------

eligibility["eligible_ct"] = (
    eligibility["ct_readable"] == True
)

eligibility["eligible_label"] = (
    eligibility["label_readable"] == True
)

eligibility["eligible_geometry"] = (
    eligibility["geometry_match"] == True
)

eligibility["eligible_label_type"] = (
    eligibility["selected_label_type"]
    .isin(["automatic", "manual"])
)

eligibility["excluded_held_out_case"] = (
    eligibility["study_id"].astype(str)
    != HELD_OUT_CASE
)

# ------------------------------------------------------------
# Final eligibility
# ------------------------------------------------------------

eligibility["eligible"] = (
    eligibility["eligible_ct"]
    & eligibility["eligible_label"]
    & eligibility["eligible_geometry"]
    & eligibility["eligible_label_type"]
    & eligibility["excluded_held_out_case"]
)

print("=" * 70)
print("BATCH 3 ELIGIBILITY")
print("=" * 70)

print(
    "Total Batch 3 candidates:",
    len(eligibility)
)

print(
    "Eligible:",
    int(eligibility["eligible"].sum())
)

print(
    "Excluded:",
    int((~eligibility["eligible"]).sum())
)
```

```python
# ============================================================
# CELL 5 — ELIGIBILITY FAILURE BREAKDOWN
# ============================================================

print("=" * 70)
print("ELIGIBILITY FAILURE BREAKDOWN")
print("=" * 70)

checks = {
    "CT unreadable":
        ~eligibility["eligible_ct"],

    "Label unreadable":
        ~eligibility["eligible_label"],

    "Geometry mismatch":
        ~eligibility["eligible_geometry"],

    "Invalid label type":
        ~eligibility["eligible_label_type"],

    "Held-out case":
        ~eligibility["excluded_held_out_case"],
}

total_failures = 0

for reason, mask in checks.items():

    count = int(mask.sum())

    print(
        f"{reason:<25}: {count}"
    )

    total_failures += count

print("\nTotal failure flags:", total_failures)

if total_failures == 0:
    print(
        "\n✓ No Batch 3 eligibility problems detected."
    )
```

```python
# ============================================================
# CELL 6 — SAVE BATCH 3 ELIGIBILITY
# ============================================================

BATCH3_ELIGIBILITY_FILE.parent.mkdir(
    parents=True,
    exist_ok=True,
)

# Save the complete eligibility table
eligibility.to_csv(
    BATCH3_ELIGIBILITY_FILE,
    index=False,
)

print("=" * 70)
print("BATCH 3 ELIGIBILITY SAVED")
print("=" * 70)

print(
    "Path:",
    BATCH3_ELIGIBILITY_FILE
)

print(
    "Rows:",
    len(eligibility)
)

print(
    "Eligible:",
    int(eligibility["eligible"].sum())
)

print(
    "Excluded:",
    int((~eligibility["eligible"]).sum())
)

print("\n✓ Batch 3 eligibility file saved.")
```

```python
# ============================================================
# CELL 7 — VERIFY SAVED ELIGIBILITY FILE
# ============================================================

check = pd.read_csv(
    BATCH3_ELIGIBILITY_FILE
)

print("=" * 70)
print("VERIFY SAVED BATCH 3 ELIGIBILITY")
print("=" * 70)

print("Rows:", len(check))

print(
    "Eligible:",
    int(check["eligible"].sum())
)

print(
    "Excluded:",
    int((~check["eligible"]).sum())
)

assert len(check) == 580
assert check["eligible"].sum() == 580
assert (~check["eligible"]).sum() == 0

print("\n✓ Saved eligibility file verified.")
```
