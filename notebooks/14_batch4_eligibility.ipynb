```python
# ============================================================
# NOTEBOOK 14 — BATCH 4 ELIGIBILITY
# CELL 1 — IMPORTS AND PATHS
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd


PROJECT_ROOT = Path(
    r"D:\Pancreatic_Cancer_Thesis"
)

DATA_DIR = PROJECT_ROOT / "data"

RAW_CT_DIR = DATA_DIR / "raw_ct"

PROCESSED_DIR = DATA_DIR / "processed"

BATCH4_INVENTORY_FILE = (
    PROCESSED_DIR / "batch4_inventory.csv"
)

BATCH4_ELIGIBILITY_FILE = (
    PROCESSED_DIR / "batch4_eligibility.csv"
)

METADATA_FILE = (
    PROCESSED_DIR / "metadata.csv"
)

HELD_OUT_CASE = "100936_00001"


print("=" * 70)
print("NOTEBOOK 14 — BATCH 4 ELIGIBILITY")
print("=" * 70)

print("\nProject root:")
print(PROJECT_ROOT)

print("\nBatch 4 inventory:")
print(BATCH4_INVENTORY_FILE)

print("\nBatch 4 eligibility:")
print(BATCH4_ELIGIBILITY_FILE)

print("\nProcessed metadata:")
print(METADATA_FILE)
```

```python
# ============================================================
# CELL 2 — LOAD BATCH 4 INVENTORY
# ============================================================

if not BATCH4_INVENTORY_FILE.exists():
    raise FileNotFoundError(
        f"Batch 4 inventory not found:\n"
        f"{BATCH4_INVENTORY_FILE}"
    )

batch4_inventory = pd.read_csv(
    BATCH4_INVENTORY_FILE
)

print("=" * 70)
print("BATCH 4 INVENTORY LOADED")
print("=" * 70)

print(
    "Rows    :",
    len(batch4_inventory)
)

print(
    "Columns :",
    len(batch4_inventory.columns)
)

print("\nColumns:")
for column in batch4_inventory.columns:
    print(" ", column)

print("\nFirst 5 cases:")
display(
    batch4_inventory.head()
)
```

```python
# ============================================================
# CELL 3 — INVENTORY SANITY CHECK
# ============================================================

print("=" * 70)
print("BATCH 4 INVENTORY SANITY CHECK")
print("=" * 70)

EXPECTED_CASES = 535

print(
    "Expected cases :",
    EXPECTED_CASES
)

print(
    "Actual cases   :",
    len(batch4_inventory)
)

assert len(batch4_inventory) == EXPECTED_CASES, (
    f"Expected {EXPECTED_CASES} cases, "
    f"found {len(batch4_inventory)}."
)

required_columns = [
    "study_id",
    "ct_readable",
    "label_readable",
    "geometry_match",
    "selected_label_type",
]

missing_columns = [
    column
    for column in required_columns
    if column not in batch4_inventory.columns
]

if missing_columns:
    raise ValueError(
        "Missing required columns:\n"
        + "\n".join(missing_columns)
    )

print("\n✓ Required columns present.")

duplicate_ids = batch4_inventory[
    batch4_inventory["study_id"].duplicated(
        keep=False
    )
]

print(
    "Duplicate study IDs:",
    len(duplicate_ids)
)

if len(duplicate_ids) > 0:
    display(duplicate_ids)
    raise ValueError(
        "Duplicate study IDs detected."
    )

print("\nCT readable:")
print(
    batch4_inventory[
        "ct_readable"
    ].value_counts(dropna=False)
)

print("\nLabel readable:")
print(
    batch4_inventory[
        "label_readable"
    ].value_counts(dropna=False)
)

print("\nGeometry match:")
print(
    batch4_inventory[
        "geometry_match"
    ].value_counts(dropna=False)
)

print("\nSelected label type:")
print(
    batch4_inventory[
        "selected_label_type"
    ].value_counts(dropna=False)
)
```

```python
# ============================================================
# CELL 4 — CHECK AGAINST EXISTING PROCESSED DATASET
# ============================================================

print("=" * 70)
print("CHECKING AGAINST EXISTING PROCESSED DATASET")
print("=" * 70)

if not METADATA_FILE.exists():
    raise FileNotFoundError(
        f"Metadata file not found:\n"
        f"{METADATA_FILE}"
    )

metadata = pd.read_csv(
    METADATA_FILE
)

processed_ids = set(
    metadata["study_id"]
    .astype(str)
    .str.strip()
)

batch4_ids = set(
    batch4_inventory["study_id"]
    .astype(str)
    .str.strip()
)

already_processed = (
    batch4_ids & processed_ids
)

new_cases = (
    batch4_ids - processed_ids
)

print(
    "Batch 4 inventory cases :",
    len(batch4_ids)
)

print(
    "Already in metadata     :",
    len(already_processed)
)

print(
    "New relative to dataset:",
    len(new_cases)
)

if already_processed:

    print("\nCases already processed:")

    for study_id in sorted(
        already_processed
    ):
        print(" ", study_id)
```

```python
# ============================================================
# CELL 5 — BUILD BATCH 4 ELIGIBILITY
# ============================================================

eligibility = batch4_inventory.copy()

eligibility["eligible_ct"] = (
    eligibility["ct_readable"] == True
)

eligibility["eligible_label"] = (
    eligibility["label_readable"] == True
)

eligibility["eligible_geometry"] = (
    eligibility["geometry_match"] == True
)

eligibility["eligible_label_type"] = (
    eligibility[
        "selected_label_type"
    ].isin(
        ["automatic", "manual"]
    )
)

eligibility["not_already_processed"] = (
    ~eligibility[
        "study_id"
    ].astype(str).isin(
        already_processed
    )
)

eligibility["not_held_out_case"] = (
    eligibility[
        "study_id"
    ].astype(str)
    != HELD_OUT_CASE
)

eligibility["eligible"] = (
    eligibility["eligible_ct"]
    &
    eligibility["eligible_label"]
    &
    eligibility["eligible_geometry"]
    &
    eligibility["eligible_label_type"]
    &
    eligibility["not_already_processed"]
    &
    eligibility["not_held_out_case"]
)

print("=" * 70)
print("BATCH 4 ELIGIBILITY")
print("=" * 70)

print(
    "Total candidates:",
    len(eligibility)
)

print(
    "Eligible:",
    int(
        eligibility["eligible"].sum()
    )
)

print(
    "Excluded:",
    int(
        (~eligibility["eligible"]).sum()
    )
)
```

```python
# ============================================================
# CELL 6 — ELIGIBILITY FAILURE BREAKDOWN
# ============================================================

print("=" * 70)
print("ELIGIBILITY FAILURE BREAKDOWN")
print("=" * 70)

checks = {
    "CT unreadable":
        ~eligibility["eligible_ct"],

    "Label unreadable":
        ~eligibility["eligible_label"],

    "Geometry mismatch":
        ~eligibility["eligible_geometry"],

    "Invalid label type":
        ~eligibility["eligible_label_type"],

    "Already processed":
        ~eligibility["not_already_processed"],

    "Held-out case":
        ~eligibility["not_held_out_case"],
}

total_failure_flags = 0

for reason, mask in checks.items():

    count = int(mask.sum())

    print(
        f"{reason:<25}: {count}"
    )

    total_failure_flags += count

print(
    "\nTotal failure flags:",
    total_failure_flags
)

if total_failure_flags == 0:
    print(
        "\n✓ No Batch 4 eligibility problems detected."
    )
```

```python
# ============================================================
# CELL 7 — CASES EXCLUDED FROM PREPROCESSING
# ============================================================

excluded = eligibility[
    ~eligibility["eligible"]
].copy()

print("=" * 70)
print("BATCH 4 EXCLUDED CASES")
print("=" * 70)

print(
    "Excluded cases:",
    len(excluded)
)

if len(excluded) > 0:

    display(
        excluded[
            [
                "study_id",
                "eligible_ct",
                "eligible_label",
                "eligible_geometry",
                "eligible_label_type",
                "not_already_processed",
                "not_held_out_case",
                "selected_label_type",
            ]
        ]
    )

else:

    print(
        "✓ No cases excluded."
    )
```

```python
# ============================================================
# CELL 8 — SAVE BATCH 4 ELIGIBILITY
# ============================================================

BATCH4_ELIGIBILITY_FILE.parent.mkdir(
    parents=True,
    exist_ok=True,
)

eligibility.to_csv(
    BATCH4_ELIGIBILITY_FILE,
    index=False,
)

print("=" * 70)
print("BATCH 4 ELIGIBILITY SAVED")
print("=" * 70)

print(
    "Path:",
    BATCH4_ELIGIBILITY_FILE
)

print(
    "Rows:",
    len(eligibility)
)

print(
    "Eligible:",
    int(
        eligibility["eligible"].sum()
    )
)

print(
    "Excluded:",
    int(
        (~eligibility["eligible"]).sum()
    )
)
```

```python
# ============================================================
# CELL 9 — VERIFY SAVED ELIGIBILITY FILE
# ============================================================

check = pd.read_csv(
    BATCH4_ELIGIBILITY_FILE
)

print("=" * 70)
print("VERIFY SAVED BATCH 4 ELIGIBILITY")
print("=" * 70)

print(
    "Rows:",
    len(check)
)

print(
    "Eligible:",
    int(
        check["eligible"].sum()
    )
)

print(
    "Excluded:",
    int(
        (~check["eligible"]).sum()
    )
)

assert len(check) == 535

assert (
    check["eligible"].sum() == 535
)

assert (
    (~check["eligible"]).sum() == 0
)

assert (
    check["study_id"]
    .astype(str)
    .nunique()
    == 535
)

print(
    "\n✓ Saved Batch 4 eligibility verified."
)
```
